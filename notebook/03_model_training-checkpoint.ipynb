# THÀNH VIÊN 3 – HUẤN LUYỆN MÔ HÌNH VÀ THỬ NGHIỆM THAM SỐ

**Đề tài:** Phân loại cảm xúc từ phản hồi của sinh viên  
**Nhiệm vụ:** Xây dựng mô hình Baseline (Naive Bayes), mô hình so sánh (K-NN), tinh chỉnh tham số và đánh giá chi tiết.

## Mục tiêu
1. Xây dựng mô hình Baseline bằng thuật toán Multinomial Naive Bayes.
2. Thử nghiệm thay đổi siêu tham số (alpha cho Naive Bayes và số láng giềng k cho K-NN).
3. Đánh giá mô hình bằng các độ đo Accuracy, Precision, Recall, F1-score và Confusion Matrix.
4. Tổng hợp và bàn giao kết quả phục vụ viết báo cáo chung của nhóm.

```python
# Import các thư viện xử lý dữ liệu và mô hình học máy
import numpy as np
import pandas as pd
from sklearn.naive_bayes import MultinomialNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

print("Đã khởi động và nạp thư viện thành công!")
```

## 1. Tải dữ liệu đã qua tiền xử lý
Nhận dữ liệu vector hóa (TF-IDF/Bag-of-Words) từ Thành viên 2 để tiến hành đưa vào huấn luyện mô hình.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split

# 1. Tạo văn bản và nhãn giả lập (mô phỏng phản hồi của sinh viên)
texts = [
    "giảng viên dạy rất nhiệt tình và dễ hiểu",
    "môn học này chán quá không hiểu gì cả",
    "bài tập rất hay và bổ ích",
    "phòng học quá nóng và chật chội",
    "tôi rất thích cách truyền đạt của thầy cô",
    "đề thi quá khó so với thời gian học"
]
labels = ["Tích cực", "Tiêu cực", "Tích cực", "Tiêu cực", "Tích cực", "Tiêu cực"]

# 2. Giả lập quá trình của Thành viên 2 (Vector hóa văn bản bằng TF-IDF)
vectorizer = TfidfVectorizer()
X_vec = vectorizer.fit_transform(texts)

# 3. Chia tập train / test giả lập
X_train_vec, X_test_vec, y_train, y_test = train_test_split(X_vec, labels, test_size=0.33, random_state=42)

print("Đã tạo thành công dữ liệu giả lập để test code!")
print("Kích thước X_train:", X_train_vec.shape)
```

## 2. Xây dựng và thử nghiệm mô hình Baseline (Multinomial Naive Bayes)
Sử dụng Multinomial Naive Bayes làm mốc so sánh chính cho bài toán xử lý ngôn ngữ tự nhiên (NLP)[cite: 7], đồng thời thực hiện khảo sát sự ảnh hưởng của tham số làm mượt `alpha`.

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# 1. Khởi tạo mô hình Naive Bayes baseline
nb_baseline = MultinomialNB()

# 2. Huấn luyện mô hình trên tập train giả lập
nb_baseline.fit(X_train_vec, y_train)

# 3. Dự đoán trên tập test
y_pred_nb = nb_baseline.predict(X_test_vec)

# 4. Đánh giá độ chính xác (Accuracy)
acc_nb = accuracy_score(y_test, y_pred_nb)
print(f"Accuracy của Naive Bayes Baseline: {acc_nb:.4f}")
```

## 3. Xây dựng và thử nghiệm mô hình so sánh (K-NN)
Thực hiện huấn luyện thuật toán K-Nearest Neighbors với các giá trị số láng giềng `k` khác nhau để so sánh hiệu năng với Naive Bayes[cite: 7].

```python
alpha_values = [0.1, 0.5, 1.0, 2.0, 5.0]
nb_results = []

for a in alpha_values:
    model = MultinomialNB(alpha=a)
    model.fit(X_train_vec, y_train)
    score = model.score(X_test_vec, y_test)
    nb_results.append({'alpha': a, 'Accuracy': score})
    print(f"Naive Bayes với alpha={a} -> Accuracy: {score:.4f}")

# Chuyển kết quả thành bảng để dễ nhìn
df_nb_results = pd.DataFrame(nb_results)
display(df_nb_results)
```

```python
from sklearn.neighbors import KNeighborsClassifier

# Tự động lấy số lượng mẫu thực tế hiện có trong tập huấn luyện
n_samples_train = X_train_vec.shape[0]

# Chọn các giá trị k an toàn (đảm bảo nhỏ hơn số lượng mẫu train)
k_values = [1, 3] 
knn_results = []

for k in k_values:
    if k < n_samples_train:
        knn_model = KNeighborsClassifier(n_neighbors=k)
        knn_model.fit(X_train_vec, y_train)
        score = knn_model.score(X_test_vec, y_test)
        knn_results.append({'k_neighbors': k, 'Accuracy': score})
        print(f"K-NN với k={k} -> Accuracy: {score:.4f}")
    else:
        print(f"Bỏ qua k={k} vì số láng giềng phải nhỏ hơn số mẫu train ({n_samples_train})")

# Chuyển kết quả thành bảng
df_knn_results = pd.DataFrame(knn_results)
display(df_knn_results)
```

## 4. Đánh giá chi tiết mô hình
Xuất báo cáo phân loại chi tiết (Classification Report) và vẽ ma trận nhầm lẫn (Confusion Matrix) để phân tích các trường hợp mô hình dự đoán đúng/sai[cite: 4, 8].

```python
from sklearn.metrics import classification_report, confusion_matrix

# In báo cáo phân loại chi tiết cho Naive Bayes (hoặc mô hình bạn muốn chọn làm mô hình chính)
print("--- BÁO CÁO PHÂN LOẠI CHI TIẾT ---")
print(classification_report(y_test, y_pred_nb))

# In ma trận nhầm lẫn
print("--- MA TRẬN NHẦM LẪN (CONFUSION MATRIX) ---")
print(confusion_matrix(y_test, y_pred_nb))
```
